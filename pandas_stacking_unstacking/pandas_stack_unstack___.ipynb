# Read a datasets 

```python
import pandas as pd
import numpy as np

np.random.seed(42)

regions = ["Kathmandu", "Pokhara", "Chitwan"]
products = ["Laptop", "Phone", "Tablet"]
months = ["Jan", "Feb", "Mar", "Apr"]

data = []

for region in regions:
    for product in products:
        for month in months:
            units = np.random.randint(20, 120)
            price = np.random.randint(300, 2000)
            sales = units * price
            data.append([region,product,month,units,price,sales])

# Create DataFrame
df = pd.DataFrame(data, columns=["Region","Product","Month","Units_Sold","Unit_Price","Total_Sales"])
df
```

```python
print(df.head())
```

## Creating a Pivots Tables 

```python
pivot_df = df.pivot_table(
    index=["Region", "Product"],
    columns="Month",
    values="Total_Sales",
    aggfunc="sum"
)

pivot_df 
```

# Stackings 

```python
pivot_df.stack()  # reset back to DataFrame
```

## Reset Index reset all the myulti headers rows 

```python
stacked_df=pivot_df.stack().reset_index()
stacked_df
```

```python
stacked_df.columns = ["Region","Product","Month","Sales"] # add dthe headings or columns 
stacked_df.head()
```

```python
print("\n================ CLEAN STACKED DATA ================\n")
stacked_df.head(10)
```

```python
unstack() converts row levels into columns

Business Use Case:
Create dashboard-ready reports
Excel-style summaries
Power BI exports
```

```python
multi_sales = (
    stacked_df
    .set_index(["Region", "Product", "Month"])
)
multi_sales
```

```python
unstacked_df = multi_sales.unstack()  # Unstack to the child level hierarchy
unstacked_df 
```

```python
unstacked_df.unstack()  # two Times Unstack convert the row level index to column level Index 
```

## Advancer unstackings 

```python
stacked_df.head()
```

### Unstack The row Index to the different Levels 

```python
advanced_unstack = (
    stacked_df
    .set_index(["Region", "Month", "Product"])
    ["Sales"]
    .unstack(level="Month")
)
advanced_unstack
```

## Unstack The data with the Region columns 

```python
advanced_unstack = (
    stacked_df
    .set_index(["Region", "Month", "Product"])
    ["Sales"]
    .unstack(level="Region")
)
advanced_unstack
```

## Unstack The data with Products 

```python
advanced_unstack = (
    stacked_df
    .set_index(["Region", "Month", "Product"])
    ["Sales"]
    .unstack(level="Product")
)
advanced_unstack
```

# Which month generated highest sales?

```python
monthly_sales = (
    stacked_df
    .groupby("Month")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
monthly_sales
```
