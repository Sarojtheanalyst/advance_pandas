# TEAM-WISE ANALYSIS

## 1. Total Runs Scored by Each Team

## 5. Team-wise Boundary Percentage

```python
df['boundary_runs'] = (
    df['4S'] * 4 +
    df['6S'] * 6
)

team_boundary = (
    df.groupby('team')
      .agg(
          total_runs=('RUNS', 'sum'),
          boundary_runs=('boundary_runs', 'sum')
      )
)

team_boundary['boundary_percentage'] = (
    team_boundary['boundary_runs'] /
    team_boundary['total_runs']
) * 100

team_boundary
```
