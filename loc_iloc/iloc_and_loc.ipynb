```python
years = [2002, 2007,1957,1962]
metrics = ["lifeExp", "pop"]
cols=[]


for item in metrics:
    for year in years:
        cols.append((item,year))
cols
```

```python
def extract_data(years,metric):
    cols=[]
    for item in metric:
        for year in years:
            cols.append((item,year))
    return temp.loc[:,cols]
    
```
