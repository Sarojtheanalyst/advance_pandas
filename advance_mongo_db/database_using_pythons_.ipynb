## extytact the dta from the databses 

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="python123",
    database="stu12"
)

cursor = conn.cursor()
```

```python
extract_query = r"""
WITH temp1 AS (
    SELECT roll_number,
           sr.question_paper_code,
           sr.question_number,
           sr.option_marked,
           ca.correct_option
    FROM student_response AS sr
    JOIN correct_answer AS ca
      ON ca.question_paper_code = sr.question_paper_code
     AND ca.question_number = sr.question_number
),

temp2 AS (
    SELECT *,
           CASE
             WHEN option_marked = 'e' THEN 'yet_to_learn'
             WHEN option_marked = correct_option THEN 'correct'
             ELSE 'incorrect'
           END AS status
    FROM temp1
),

temp3 AS (
    SELECT *
    FROM temp2 AS t2
    JOIN question_paper_code AS qc
      ON t2.question_paper_code = qc.paper_code
),

temp4 AS (
    SELECT roll_number,
           subject,
           status,
           COUNT(status) AS correct_incorrect
    FROM temp3
    GROUP BY roll_number, subject, status
),

temp5 AS (
    SELECT *,
           CONCAT(subject, '_', status) AS combined
    FROM temp4
),

temp6 AS (
    SELECT roll_number,
           SUM(CASE WHEN combined = 'science_incorrect' THEN correct_incorrect ELSE 0 END) AS science_incorrect,
           SUM(CASE WHEN combined = 'science_yet_to_learn' THEN correct_incorrect ELSE 0 END) AS science_yet_to_learn,
           SUM(CASE WHEN combined = 'science_correct' THEN correct_incorrect ELSE 0 END) AS science_correct,
           SUM(CASE WHEN combined = 'math_correct' THEN correct_incorrect ELSE 0 END) AS math_correct,
           SUM(CASE WHEN combined = 'math_incorrect' THEN correct_incorrect ELSE 0 END) AS math_incorrect,
           SUM(CASE WHEN combined = 'math_yet_to_learn' THEN correct_incorrect ELSE 0 END) AS math_yet_to_learn
    FROM temp5
    GROUP BY roll_number
),

temp7 AS (
    SELECT *,
           ROUND((science_correct / (science_incorrect + science_yet_to_learn + science_correct)) * 100, 2) AS science_per,
           ROUND((math_correct / (math_correct + math_incorrect + math_yet_to_learn)) * 100, 2) AS math_per
    FROM temp6
)

SELECT *
FROM student_list AS sl
JOIN temp7 AS t7
  ON t7.roll_number = sl.roll_number
"""
```

```python
cursor.execute(extract_query)

for row in cursor.fetchall():
    print(row)
```
