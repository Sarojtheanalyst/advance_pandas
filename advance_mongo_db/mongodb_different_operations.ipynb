# 81. Complex Aggregation: Movies with the Highest Gross in Each Genre 

```python
highest_gross_movies_per_genre = collection.aggregate([ 
{"$unwind": "$Genre"}, 
{"$group": { 
"_id": "$Genre", 
"highest_gross": {"$max": "$Gross"}, 
"movie_details": {"$first": "$$ROOT"}  # This gets the whole document of the movie 
}}, 
{"$sort": {"highest_gross": -1}} 
]) 
for genre in highest_gross_movies_per_genre: 
    print(genre)
```
