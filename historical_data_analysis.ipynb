```python
cols = [
    'Open','High','Low','Close','VWAP',
    'Vol','Prev. Close','Turnover',
    '120 Days','180 Days',
    '52 Weeks High','52 Weeks Low'
]

for col in cols:
    combined_df[col] = pd.to_numeric(
        combined_df[col]
          .astype(str)
          .str.replace(',', '', regex=False)
          .replace('-', np.nan),
        errors='coerce'
    )
```

```python
combined_df['Date'] = pd.to_datetime(
    combined_df['level_0'],
    format='%Y_%m_%d'
)
```

```python
combined_df = combined_df.sort_values(
    ['Symbol', 'Date']
)
```

```python
# future returns targets

combined_df = combined_df.sort_values(["Symbol", "Date"])

for days in [5, 10, 20]:
    combined_df[f"future_close_{days}"] = (
        combined_df.groupby("Symbol")["Close"]
          .shift(-days)
    )

    combined_df[f"future_return_{days}"] = (
        combined_df[f"future_close_{days}"] / combined_df["Close"] - 1
    )
```

```python
combined_df.head(30)
```

```python
# near 52 wewk high
combined_df["near_high"] = (
    combined_df["Close"] / combined_df["52 Weeks High"]
)
```

```python
# above Moving averages
combined_df["above120"] = (
    combined_df["Close"] / combined_df["120 Days"]
)

combined_df["above180"] = (
    combined_df["Close"] / combined_df["180 Days"]
)
```

```python
combined_df["range_pct"] = (
    (combined_df["High"] - combined_df["Low"])
    / combined_df["Close"]
)
```

```python
# volume surge
combined_df["avg_vol20"] = (
    combined_df.groupby("Symbol")["Vol"]
      .transform(lambda x: x.rolling(20).mean())
)

combined_df["vol_ratio"] = (
    combined_df["Vol"] / combined_df["avg_vol20"]
)
```

```python
# Turnover surge

combined_df["avg_turn20"] = (
    combined_df.groupby("Symbol")["Turnover"]
      .transform(lambda x: x.rolling(20).mean())
)

combined_df["turnover_ratio"] = (
    combined_df["Turnover"] / combined_df["avg_turn20"]
)
```


# Top 5 Gainers and Losers




```python
def extract_top_5_gainers(date):

    day_data = combined_df[combined_df["Date"] == date].copy()

    gainers = (day_data.nlargest(5, "Diff %")[["Symbol", "Diff %"]].sort_values("Diff %"))
    losers = (day_data.nsmallest(5, "Diff %")[["Symbol", "Diff %"]].sort_values("Diff %"))
    fig, axes = plt.subplots(1, 2, figsize=(15, 4))

    # Top Gainers
    sns.barplot(data=gainers,x="Diff %",y="Symbol",ax=axes[0],color="green")
    axes[0].set_title(f"Top 5 Gainers ({date})")
    axes[0].set_xlabel("Diff %")

    for i, v in enumerate(gainers["Diff %"]):
        axes[0].text(v,i,f"{v:.2f}%",va="center")

    # Top Losers
    sns.barplot(data=losers,x="Diff %",y="Symbol",ax=axes[1],color="red")
    axes[1].set_title(f"Top 5 Losers ({date})")
    axes[1].set_xlabel("Diff %")
    for i, v in enumerate(losers["Diff %"]):
        axes[1].text(v,i,f"{v:.2f}%",va="center")

    plt.tight_layout()
    plt.show()

    return gainers, losers
```

```python
def plot_smoothed_series(
    df,
    date_col="Date",
    value_col="Close",
    window=20,
    title=None
):

    plot_df = df.copy()

    plot_df[date_col] = pd.to_datetime(
        plot_df[date_col],
        format="%Y_%m_%d"
    )

    plot_df = plot_df.sort_values(date_col)

    # Moving Average
    plot_df["MA"] = (
        plot_df[value_col]
        .rolling(window=window)
        .mean()
    )

    fig, ax = plt.subplots(figsize=(15,6))

    # Original
    ax.plot(
        plot_df[date_col],
        plot_df[value_col],
        alpha=0.7,
        label="Original"
    )

    # Smoothed
    ax.plot(
        plot_df[date_col],
        plot_df["MA"],
        linewidth=3,
        label=f"{window}-Day MA"
    )

    ax.xaxis.set_major_locator(
        mdates.MonthLocator(interval=1)
    )

    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%Y-%m")
    )

    plt.xticks(rotation=45)

    plt.title(
        title if title else value_col
    )

    plt.legend()
    plt.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()
```

# Convert to weekly analysis

```python
market = (
    combined_df
    .groupby("Date")
    .agg(
        Avg_Return=("Diff %","mean"),
        Total_Turnover=("Turnover","sum"),
        Total_Volume=("Vol","sum")
    )
    .reset_index()
)

market["Date"] = pd.to_datetime(
    market["Date"],
    format="%Y_%m_%d"
)
market
```

# Create Weekly Data

```python
weekly = (
    market
    .set_index("Date")
    .resample("W")
    .agg({
        "Avg_Return":"mean",
        "Total_Turnover":"sum",
        "Total_Volume":"sum"
    })
    .reset_index()
)
weekly
```
