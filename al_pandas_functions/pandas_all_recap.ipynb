```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

np.random.seed(42)
```

```python
n = 50000

df = pd.DataFrame({

    "order_id": np.arange(1, n+1),

    "customer_id": np.random.randint(1000, 10000, n),

    "customer_name": np.random.choice(
        ["Ram", "Hari", "Sita", "John", "Emily", "David"],
        n
    ),

    "gender": np.random.choice(
        ["Male", "Female"],
        n
    ),

    "city": np.random.choice(
        ["Kathmandu", "Pokhara", "Lalitpur", "Biratnagar"],
        n
    ),

    "department": np.random.choice(
        ["Electronics", "Fashion", "Food", "Furniture"],
        n
    ),

    "sales_channel": np.random.choice(
        ["Online", "Retail", "Distributor"],
        n
    ),

    "payment_method": np.random.choice(
        ["Cash", "Card", "eSewa", "Khalti"],
        n
    ),

    "quantity": np.random.randint(1, 10, n),

    "unit_price": np.random.randint(100, 5000, n),

    "discount": np.random.uniform(0, 0.40, n),

    "rating": np.random.uniform(1, 5, n),

    "delivery_days": np.random.randint(1, 15, n),

    "website_visits": np.random.randint(1, 500, n),

    "ad_spend": np.random.randint(1000, 50000, n),

    "join_date": pd.to_datetime("2020-01-01") +
                 pd.to_timedelta(
                     np.random.randint(0, 1500, n),
                     unit="D"
                 ),

    "order_date": pd.to_datetime("2023-01-01") +
                  pd.to_timedelta(
                      np.random.randint(0, 730, n),
                      unit="D"
                  )
})

df["sales"] = (
    df["quantity"] *
    df["unit_price"] *
    (1 - df["discount"])
)

df["profit"] = (
    df["sales"] *
    np.random.uniform(0.05, 0.30, n)
)

df.head()
```
