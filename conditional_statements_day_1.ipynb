```python
print(f'''
1) $ (Dollar) - ₹ (Indian Rupee)
2) Nrs (nepal) - $ (Dollar): 
''')

choice=input("Ebter The Choice : ")

if choice=='1':
    print(f'''
       a) $ -  ₹
       b) ₹- $
     ''')
    
    c=input("Enter Your Choice : ")
    
    if c=='a':
        dollar=int(input("Enter Yhe ASmount in ($) "))
        inr=round(94.53*dollar,2)
        print(f'${dollar} =  ₹ {inr}')
        
    elif c=='b':
        inr=int(input("Enter The Amount  in (₹) "))
        dolllar=round(inr/94.53,2)
        print(f'₹{inr} =  ${dolllar}')
        
    else:
        print("invalid Input ")

elif choice=='2':
    
    print(f'''
       a) Nrs -  $
       b) $- Nrs
    ''')
    
    c=input("Enter Your Choice : ")
    
    if c=='a':
        nrs=int(input("Enter Yhe ASmount in (Nrs) "))
        dollar=round(nrs/151.14 ,2)
        print(f'Nrs. {nrs} =  $ { dollar}')
        
    elif c=='b':
        dollar=int(input("Enter The Amount  in ($) "))
        nrs=round(dollar*151.14 ,2)
        print(f'${dollar} =  Nrs. {nrs}')
        
    else:
        print("Invalid Input ")

else:
    print("Invalid input ")
    
```
