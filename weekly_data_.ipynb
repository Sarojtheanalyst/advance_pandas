```python
def data_converter(data, sector):
    rows = []
    for item in data.split("\n")[1:]:
        date = item.split()[1]
        pct_changes = item.split()[-1]

        rows.append([date, pct_changes])

    df = pd.DataFrame(rows, columns=["Date", "pct_changes"])
    df["sector"] = sector
    return df
```
