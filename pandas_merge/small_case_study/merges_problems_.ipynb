# 1) Find The Total Revenue Generated : jan , feb 

```python
temp=pd.concat([jan,feb],keys=["jan",'feb'])
temp=temp.reset_index()

temp.merge(courses,on='course_id',how='inner')['price'].sum()
```

# 2) Find The Total Revenue Genretaed by the course in each Month (jan, feb)

```python
temp=pd.concat([jan,feb],keys=["jan",'feb']).reset_index()
temp.merge(courses,on="course_id",how="inner").groupby('level_0')['price'].sum()
```
