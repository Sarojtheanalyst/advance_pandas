```python


# =========================================================
# INNER MERGE
# =========================================================
# Merge Orders with Customers

inner_merge = pd.merge(
    orders,
    customers,
    on="customer_id",
    how="inner"
)

print("\nINNER MERGE")
print(inner_merge.head())


# =========================================================
# LEFT MERGE
# =========================================================
# Merge Orders with Payments

left_merge = pd.merge(
    orders,
    payments,
    on="order_id",
    how="left"
)

print("\nLEFT MERGE")
print(left_merge.head())


# =========================================================
# MULTIPLE TABLE MERGE
# =========================================================

full_data = (
    orders
    .merge(customers, on="customer_id", how="left")
    .merge(products, on="product_id", how="left")
    .merge(payments, on="order_id", how="left")
)

print("\nFULL MERGED DATASET")
print(full_data.head())


# =========================================================
# CALCULATE TOTAL SALES
# =========================================================

full_data["total_sales"] = (
    full_data["quantity"] * full_data["price"]
)

print("\nFINAL DATASET WITH SALES")
print(full_data.head())


# =========================================================
# BUSINESS ANALYSIS
# =========================================================

sales_summary = (
    full_data.groupby("category")["total_sales"]
    .sum()
    .reset_index()
    .sort_values(by="total_sales", ascending=False)
)

print("\nSALES SUMMARY")
print(sales_summary)
```
