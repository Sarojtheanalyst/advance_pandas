## Implement in real data 

```python
df=pd.read_csv(r"C:\Users\HP\Desktop\gap.csv")
df
```

```python
summary=pd.pivot(df,columns="year", index="country", values="pop").loc[["Nepal","India","China"]]
summary
```

```python
current=np.array(summary)[:,1:]
current
```

```python
previous=np.array(summary)[:,:-1]
previous
```

```python
np.round((current-previous)/(previous)*100,2)
```

```python
curr=np.array(summary.columns).astype(str)[1:]
prev=np.array(summary.columns).astype(str)[:-1]
```

```python
curr
```

```python
prev
```

```python
final_cols=prev+"_"+curr
final_cols
```

```python
summary
```

```python
pd.DataFrame(np.round((current-previous)/(previous)*100,2),index=summary.index,columns=final_cols)
```
