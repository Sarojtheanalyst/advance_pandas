```python
import time


for url in closed_ended[40:]:
    driver.get(url)
    time.sleep(5)
    weekly_nav=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[1]/div[1]/div[3]").text
    schema_profiles=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[1]/div[2]/div/button/div").text
    corporate_actions=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[2]/div[3]/div").text
    market_summary=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[3]/div[1]/div[2]").text
    performance=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[3]/div[2]/div/table").text
    financials=driver.find_element(By.XPATH,"/html/body/div[7]/div[1]/div/div/div[3]/div[3]/div/table").text
    assets_allocations=driver.find_element(By.XPATH,"/html/body/div[7]/div[2]/div/div/div[2]/div").text
    sectorwise_allocations=driver.find_element(By.XPATH,"/html/body/div[7]/div[2]/div/div/div[3]/div/div[3]").text
    company_id=url.split("=")[1]   

    info_dicts={"weekly_nav":weekly_nav,"schema_profiles":schema_profiles,"corporate_actions":corporate_actions,
     "market_summary":market_summary,"performance":performance,"financials":financials,
      "assets_allocations":assets_allocations,"sectorwise_allocations":sectorwise_allocations}
    all_dicts[company_id]=info_dicts
        
```
