```python
from selenium import webdriver
import json

options = webdriver.ChromeOptions()
options.set_capability("goog:loggingPrefs", {"performance": "ALL"})
driver = webdriver.Chrome(options=options)
driver.get("https://app.smartwealthpro.com")
input("👉 Interact with website then press ENTER...")
logs = driver.get_log("performance")
apis = set()

for entry in logs:
    try:
        message = json.loads(entry["message"])["message"]
        if message.get("method") == "Network.requestWillBeSent":
            params = message.get("params", {})
            request = params.get("request", {})
            url = request.get("url")
            if url and "/api/" in url:
                apis.add(url)

    except Exception:
        pass

print("\n=== FOUND APIs ===\n")

for api in sorted(apis):
    print(api)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def fetch_api_datadset(urls):
    login_url = "https://app.smartwealthpro.com/Account/Login"
    api_url = urls
    USERNAME = "muskasturid@gmail.com"
    PASSWORD = "Nep@l2053"
    session = requests.Session()
    headers = {
        "User-Agent": "Mozilla/5.0",
        "Referer": login_url
    }
    response = session.get(login_url, headers=headers)
    print("LOGIN PAGE STATUS:", response.status_code)
    soup = BeautifulSoup(response.text, "html.parser")
    token = soup.find(
        "input",
        {"name": "__RequestVerificationToken"}
    )["value"]
    
    print("TOKEN FOUND")
    payload = {
        "__RequestVerificationToken": token,
        "UserName": USERNAME,
        "Password": PASSWORD,
        "RememberMe": "false"
    }
    login_response = session.post(
        login_url,
        data=payload,
        headers=headers
    )
    print("LOGIN STATUS:", login_response.status_code)
    if "Log in" in login_response.text:
        print("LOGIN FAILED")
    else:
        print("LOGIN SUCCESS")
    
    # STEP 5: ACCESS API
    api_headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept": "application/json, text/plain, */*",
        "Referer": "https://app.smartwealthpro.com/"
    }
    api_response = session.get(
        api_url,
        headers=api_headers
    )
    print("API STATUS:", api_response.status_code)
    try:
    
        data = api_response.json()
        print("JSON SUCCESS")
        df = pd.DataFrame(data)
        return df
        # df.to_csv(f"C:\\Users\\HP\\Desktop\\data_extractions\\extraced_datasets\\data_fetch_of_all_comoany\\all_final_datasets\\company_\\{folder_name}\\{end_points}.csv", index=False)
    except:
        pass
```
