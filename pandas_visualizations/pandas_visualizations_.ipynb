# Import The Library 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')

print(" Imported the Library sucessfully ")
```

```python
df=pd.read_csv("sales.csv")
df
```

```python
df.set_index("Date", inplace=True)
```

```python
df.head()
```

# 1. SALES TREND (LINE CHART – CLEAN & MODERN)

```python
df["Sales"].plot(kind="line",figsize=(14,6),title="Daily Sales Trend",color="green",linewidth=2)
plt.grid(alpha=0.3)
plt.show()
```

```python
📊 1. LINE CHART (kind="line")

📈 What it does: Shows trend over time

✔ Key Points:
Best for time-series data
Shows rise/fall patterns clearly
Helps detect seasonality and trends

💼 Use Cases:
Daily sales tracking
Stock market price movement
Website traffic over time
```

# 2. AREA CHART (BUSINESS FLOW)

```python
df["Sales"].plot(kind="area",figsize=(14,6),title="Sales Area Trend",color="#A1C9F4",alpha=0.6)
plt.grid(alpha=0.3)
plt.show()
```

```python
📊 2. AREA CHART (kind="area")
🌊 What it does: Shows cumulative trend with filled area

✔ Key Points:
Emphasizes volume over time
Better visual impact than line chart
Can show stacked contribution (multi-series)

💼 Use Cases:
Revenue flow over time
Resource usage trends
Cumulative growth analysis
```

# 3. HISTOGRAM (DISTRIBUTION)

```python
df["Sales"].plot(kind="hist",bins=np.arange(3000,8000,500),figsize=(10,5),title="Sales Distribution",
                 color="red",edgecolor="white")
plt.xticks(np.arange(3000,8000,500))
plt.grid(alpha=0.2)
plt.show()
```

```python
📦 3. HISTOGRAM (kind="hist")

📊 What it does: Shows frequency distribution

✔ Key Points:
Groups data into bins
Shows distribution shape
Helps detect skewness

💼 Use Cases:
Salary distribution
Sales distribution
Exam score analysis
```

# 4. KDE / DENSITY (SMOOTH DISTRIBUTION)

```python
df["Sales"].plot(kind="density",figsize=(10,5),title="Sales Density Curve",color="#FFB5A7",linewidth=2)
plt.grid(alpha=0.2)
plt.show()
```

```python
📉 4. KDE / DENSITY (kind="density")

📉 What it does: Smooth version of histogram

✔ Key Points:
Shows probability distribution shape
Removes bin noise from histogram
Helps compare distributions

💼 Use Cases:
Customer spending behavior
Risk distribution in finance
Data normality check
```

# 5. BOX PLOT (OUTLIERS DETECTION)

```python
df[["Sales","Visitors"]].plot(kind="box",figsize=(10,5),title="KPI Spread & Outliers")
plt.grid(alpha=0.2)
plt.show()
```

```python
📦 5. BOX PLOT (kind="box")
📊 What it does: Shows spread, median, and outliers

✔ Key Points:
Shows 5-number summary
Detects outliers clearly
Compares multiple groups easily

💼 Use Cases:
Salary comparison across departments
Sales variation across regions
Quality control in manufacturing
```

# 6. MULTI-LINE (BUSINESS DASHBOARD VIEW)

```python
df[["Orders","Visitors"]].plot(kind="line",figsize=(14,6),title="Business KPI Trend",linewidth=2
)
plt.grid(alpha=0.3)
plt.show()
```

# 7. SCATTER (RELATIONSHIP ANALYSIS)

```python
df.plot(kind="scatter",x="Visitors",y="Sales",figsize=(8,6),title="Visitors vs Sales",color="green"
)
plt.grid(alpha=0.3)
plt.show()
```

```python
🔗 6. SCATTER PLOT (kind="scatter")

🔵 What it does: Shows relationship between 2 variables

✔ Key Points:
Shows correlation visually
Helps detect patterns or clusters
Useful for regression analysis

💼 Use Cases:
Visitors vs Sales relationship
Height vs Weight analysis
Ad spend vs revenue
```

# 8. BAR (MONTHLY AGGREGATION)

```python
df["Sales"].resample("M").sum().plot(kind="barh",figsize=(10,5),title="Monthly Sales Performance",
                                     color="#64B5F6")
plt.grid(alpha=0.2)
plt.show()
```

```python
📊 7. BAR CHART (kind="bar")

📊 What it does: Compares categories

✔ Key Points:
Best for categorical comparison
Easy ranking visualization
Works on aggregated data
    
💼 Use Cases:
Product sales comparison
Region-wise performance
Monthly revenue comparison
```

# 9 ) PIE CHART Comparisions)

```python
df_sum = df[["Sales", "Orders", "Visitors"]].mean()
df_sum
```

```python
df_sum.plot(kind="pie",figsize=(7,10),autopct="%1.1f%%",startangle=90,
            title="Business KPI Contribution Share")

plt.ylabel("")
plt.show()
```

```python
🥧 9. PIE CHART (kind="pie")

🥧 What it does: Shows proportion of whole

✔ Key Points:
Shows percentage contribution
Works best with few categories
Not suitable for time-series data

💼 Use Cases:
Market share analysis
Budget allocation
KPI contribution breakdown
```
