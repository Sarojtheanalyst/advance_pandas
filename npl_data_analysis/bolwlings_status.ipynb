# Question 2 : Highest Wicket takers in NPL

```python
final["dismissal_kind"] = (
    final["dismissal_kind"]
    .str.lower()
    .str.strip()
)
```

# sql equivalents 

```python
SELECT
    bowler,
    COUNT(*) AS wickets,
    SUM(CASE WHEN dismissal_kind IN ('caught') THEN 1 ELSE 0 END) AS caught,
    SUM(CASE WHEN dismissal_kind IN ('bowled') THEN 1 ELSE 0 END) AS bowled,
    SUM(CASE WHEN dismissal_kind IN ('lbw') THEN 1 ELSE 0 END) AS lbw,
    SUM(CASE WHEN dismissal_kind IN ('stumped') THEN 1 ELSE 0 END) AS stumped,
    SUM(CASE WHEN dismissal_kind IN ('caught & bowled') THEN 1 ELSE 0 END) AS caught_bowled,
    SUM(CASE WHEN dismissal_kind IN ('hit wicket') THEN 1 ELSE 0 END) AS hit_wicket
FROM npl_ball_by_ball

WHERE dismissal_kind IS NOT NULL
AND dismissal_kind NOT IN (
    'runout',
    'run out',
    'retired hurt'
)
GROUP BY bowler
ORDER BY wickets DESC;
```

### pandas equivalents 

```python
valid_wickets = final[
    (~final["dismissal_kind"].isin(
        ["runout", "run out", "retired hurt"]
    )) &
    (final["dismissal_kind"].notna())
]

result = (
    valid_wickets
    .groupby("bowler")
    .agg(
        wickets=("dismissal_kind", "count"),
        caught=("dismissal_kind",lambda x: (x == "caught").sum()),
        bowled=("dismissal_kind",lambda x: (x == "bowled").sum()),
        lbw=("dismissal_kind",lambda x: (x == "lbw").sum()),
        stumped=("dismissal_kind",lambda x: (x == "stumped").sum()),
        caught_bowled=("dismissal_kind",lambda x: (x == "caught & bowled").sum()),
        hit_wicket=("dismissal_kind",lambda x: (x == "hit wicket").sum())
    )
    .reset_index()
    .sort_values("wickets", ascending=False)
)

result.head(10)
```

# Question 3 :  Highest wicket Talen By the bowler in an NPL matches

### sql queries 

```python
SELECT
    match_id,
    bowler,
    COUNT(*) AS wickets,
    SUM(total_runs) AS runs_conceded,
    CONCAT(
        COUNT(*),
        '-',
        SUM(total_runs)
    ) AS best_figures

FROM npl_ball_by_ball
WHERE dismissal_kind IS NOT NULL
AND LOWER(dismissal_kind) NOT IN (
    'runout',
    'run out',
    'retired hurt'
)

GROUP BY match_id, bowler
ORDER BY wickets DESC,
         runs_conceded ASC;
```

## pandas equivalents 

```python
valid = final[
    (final["dismissal_kind"].notna()) &
    (~final["dismissal_kind"].isin(
        ["runout", "run out", "retired hurt"]
    ))
]

wickets = (
    valid.groupby(["match_id", "bowler"])
    .size()
    .reset_index(name="wickets")
)

runs = (
    final.groupby(["match_id", "bowler"])["total_runs"]
      .sum()
      .reset_index(name="runs")
)

result = wickets.merge(
    runs,
    on=["match_id", "bowler"]
)

result["best_figures"] = (
    result["wickets"].astype(str)
    + "-"
    + result["runs"].astype(str)
)

result = result.sort_values(
    ["wickets", "runs"],
    ascending=[False, True]
)

result.head()
```

#  Question No 4 : No of 5-wicket hauls by bowlers in an NPL

```python
WITH match_wickets AS (
    SELECT
        match_id,
        bowler,
        COUNT(*) AS wickets
    FROM npl_ball_by_ball
    WHERE dismissal_kind IS NOT NULL
    AND LOWER(dismissal_kind) NOT IN (
        'runout',
        'run out',
        'retired hurt'
    )
    GROUP BY match_id, bowler
)

SELECT
    bowler,
    COUNT(*) AS five_wicket_hauls
FROM match_wickets
WHERE wickets >= 5
GROUP BY bowler
ORDER BY five_wicket_hauls DESC;
```

```python
final["dismissal_kind"] = (
    final["dismissal_kind"]
      .str.lower()
      .str.strip()
)

valid_wickets = final[
    (final["dismissal_kind"].notna()) &
    (~final["dismissal_kind"].isin(
        ["runout", "run out", "retired hurt"]
    ))
]

match_wickets = (
    valid_wickets
    .groupby(["match_id", "bowler"])
    .size()
    .reset_index(name="wickets")
)

five_wicket_hauls = (
    match_wickets[
        match_wickets["wickets"] >= 5
    ]
    .groupby("bowler")
    .size()
    .reset_index(name="hauls")
    .sort_values("hauls", ascending=False)
)

five_wicket_hauls
```

# Question No 7 : Best Economy And Worst Economy  on NPL 

```python
WITH bowler_stats AS (
    SELECT
        bowler,

        COUNT(*) AS balls,

        ROUND(COUNT(*) / 6, 1) AS overs,

        SUM(total_runs) AS runs,

        SUM(wide + noball) AS extras,

        ROUND(
            SUM(total_runs) /
            NULLIF(COUNT(*) / 6, 0),
        2) AS economy

    FROM npl_ball_by_ball

    GROUP BY bowler

    HAVING COUNT(*) >= 300
)

SELECT
    bowler,
    balls,
    overs,
    runs,
    extras,
    economy

FROM bowler_stats

ORDER BY economy ASC;
```

```python
WITH death_overs AS (
    SELECT
        bowler,
        match_id,
        FLOOR(over_ball) AS over_no,
        total_runs,
        dismissal_kind

    FROM npl_ball_by_ball

    WHERE FLOOR(over_ball) IN (16,17,18,19,20)
),

bowler_stats AS (
    SELECT
        bowler,

        COUNT(*) / 6 AS overs,

        SUM(total_runs) AS runs,

        ROUND(
            SUM(total_runs) /
            NULLIF(COUNT(*) / 6, 0),
        2) AS economy

    FROM death_overs

    GROUP BY bowler
),

wickets AS (
    SELECT
        bowler,

        COUNT(*) AS wickets

    FROM death_overs

    WHERE LOWER(dismissal_kind) IN (
        'caught',
        'bowled',
        'lbw',
        'stumped',
        'caught & bowled',
        'hit wicket'
    )

    GROUP BY bowler
)

SELECT
    bs.bowler,
    ROUND(bs.overs,1) AS overs,
    bs.runs,

    COALESCE(w.wickets,0) AS wickets,

    bs.economy,

    ROUND(
        COALESCE(w.wickets,0) /
        NULLIF(bs.overs,0),
    2) AS wicket_rate

FROM bowler_stats bs

LEFT JOIN wickets w
    ON bs.bowler = w.bowler

WHERE bs.overs >= 10

ORDER BY
    economy ASC,
    wicket_rate DESC;
```

# Question 13 : Best Bowlers and Worst Bowlers in a NPL

```python
WITH bowler_runs AS (
    SELECT
        bowler,

        COUNT(*) AS balls,

        ROUND(COUNT(*) / 6,1) AS overs,

        SUM(total_runs) AS runs,

        ROUND(
            SUM(total_runs) /
            NULLIF(COUNT(*) / 6,0),
        2) AS economy

    FROM npl_ball_by_ball

    GROUP BY bowler

    HAVING COUNT(*) >= 300
),

wickets AS (

    SELECT
        bowler,
        COUNT(*) AS wickets

    FROM npl_ball_by_ball

    WHERE LOWER(TRIM(dismissal_kind)) IN (
        'caught',
        'bowled',
        'lbw',
        'stumped',
        'caught & bowled',
        'hit wicket'
    )

    GROUP BY bowler
),

matches AS (

    SELECT
        bowler,
        COUNT(DISTINCT match_id) AS matches

    FROM npl_ball_by_ball

    GROUP BY bowler
),

best_figures AS (

    SELECT
        bowler,

        MAX(
            CONCAT(
                wickets,
                '-',
                runs
            )
        ) AS best

    FROM (

        SELECT
            match_id,
            bowler,

            COUNT(*) AS wickets,

            SUM(total_runs) AS runs

        FROM npl_ball_by_ball

        WHERE LOWER(TRIM(dismissal_kind)) IN (
            'caught',
            'bowled',
            'lbw',
            'stumped',
            'caught & bowled',
            'hit wicket'
        )

        GROUP BY
            match_id,
            bowler

    ) x

    GROUP BY bowler
)

SELECT
    r.bowler,
    m.matches,
    r.runs,
    w.wickets,
    r.economy,
    b.best

FROM bowler_runs r

JOIN wickets w
    ON r.bowler = w.bowler

JOIN matches m
    ON r.bowler = m.bowler

JOIN best_figures b
    ON r.bowler = b.bowler

ORDER BY
    economy ASC,
    wickets DESC;
```
