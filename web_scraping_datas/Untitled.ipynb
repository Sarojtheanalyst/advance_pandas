# Extract the tables datas 

```python
import pandas as pd
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

# Open Chrome
driver = webdriver.Chrome()

# Open URL
driver.get("https://www.sharesansar.com/proposed-dividend")

# Wait for loading
time.sleep(5)

# Find table
table = driver.find_element(By.TAG_NAME, "table")

# Extract rows
rows = table.find_elements(By.TAG_NAME, "tr")

data = []

for row in rows:

    cols = row.find_elements(By.TAG_NAME, "td")

    if not cols:
        cols = row.find_elements(By.TAG_NAME, "th")

    row_data = [col.text.strip() for col in cols]

    if row_data:
        data.append(row_data)

# Close browser
driver.quit()

# Create dataframe
headers = data[0]
table_data = data[1:]

df = pd.DataFrame(table_data, columns=headers)

print(df)

# # Save CSV
# df.to_csv("proposed_dividend.csv", index=False)
```

## fins how many oages are there 

```python
import re
import json
from selenium import webdriver

driver = webdriver.Chrome()
driver.get("https://www.sharesansar.com/proposed-dividend")

html = driver.page_source

# Extract cmpjson array
match = re.search(r"var cmpjson = (\[.*?\]);", html, re.DOTALL)

if match:
    json_text = match.group(1)
    data = json.loads(json_text)

    for item in data:
        print(item)

else:
    print("cmpjson not found")

driver.quit()
```
