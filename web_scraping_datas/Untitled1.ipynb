## Automations 

```python
from apscheduler.schedulers.background import BackgroundScheduler
import time
import sys

def job():
    print("Running job...")

scheduler = BackgroundScheduler()
scheduler.add_job(job, 'interval', seconds=2)
scheduler.start()

print("Started... Press CTRL + C to stop")

try:
    while True:
        time.sleep(1)

except KeyboardInterrupt:
    scheduler.shutdown()
    sys.exit(0)
```
