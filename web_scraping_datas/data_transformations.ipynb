```python
def transform(temp):
    temp=np.array(temp)
    cols=np.unique(temp[:,0:1])
    final_arr=[]
    for arr in np.vsplit(temp[:,1:],temp.shape[0]//3):
        final_arr.append(arr.T)
    final_dfs=pd.DataFrame(np.concat(final_arr),columns=cols)
    return final_dfs
    
```
