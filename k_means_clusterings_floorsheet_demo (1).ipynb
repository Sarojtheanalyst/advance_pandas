## calculate the Buy Broker 

```python
buy = final_dfs.groupby("Buyer Broker").agg(
    total_buy_qty=("Share Quantity", "sum"),
    total_buy_amt=("Amount (in Rs)", "sum"),
    buy_txns=("Transaction No.", "count")
).reset_index()

buy = buy.rename(columns={"Buyer Broker": "broker_no"})
buy
```

```python
sell = final_dfs.groupby("Sell Broker").agg(
    total_sell_qty=("Share Quantity", "sum"),
    total_sell_amt=("Amount (in Rs)", "sum"),
    sell_txns=("Transaction No.", "count")
).reset_index()

sell = sell.rename(columns={"Sell Broker": "broker_no"})
sell
```

```python
broker = pd.merge(buy, sell, on="broker_no", how="outer").fillna(0)
broker
```

# Calculatings the different metrics 

```python
broker["total_turnover"] = broker["total_buy_amt"] + broker["total_sell_amt"]
broker["net_qty"] = broker["total_buy_qty"] - broker["total_sell_qty"]
broker["net_amt"] = broker["total_buy_amt"] - broker["total_sell_amt"]
broker["buy_sell_ratio"] = (
    broker["total_buy_amt"] / broker["total_sell_amt"].replace(0, pd.NA)
)
broker
```

```python
broker = broker[
    [
        "broker_no",
        "total_turnover",
        "total_buy_qty",
        "total_buy_amt",
        "buy_txns",
        "total_sell_qty",
        "total_sell_amt",
        "sell_txns",
        "net_qty",
        "net_amt",
        "buy_sell_ratio"
    ]
]
broker 
```

## Implement the k means algorithms 

```python
features = ["total_buy_qty","total_sell_qty","total_buy_amt","total_sell_amt","buy_txns","sell_txns",
            "net_qty","net_amt","total_turnover"]
df_cluster = broker[features].fillna(0)
df_cluster
```

## standasrd scalars 

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df_cluster)
scaled_data
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
```

```python
inertia = []
silhouette_scores = []

K_range = range(2, 11)
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(scaled_data)
    inertia.append(kmeans.inertia_)
    score = silhouette_score(scaled_data, labels)
    silhouette_scores.append(score)
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

### plot to show the best fit numbers of k 

```python
plt.figure(figsize=(8,5))
plt.plot(K_range, silhouette_scores, marker='o', color='green')
plt.title("Silhouette Score vs K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.grid()
plt.show()
```

## Creating a 2 clusters 

```python
means = KMeans(n_clusters=2, random_state=42, n_init=10)

df_cluster['Cluster'] = means.fit_predict(scaled_data)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
components = pca.fit_transform(scaled_data)

df_cluster['PCA1'] = components[:, 0]
df_cluster['PCA2'] = components[:, 1]
```
